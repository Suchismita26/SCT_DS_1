```python
population_2022 = df[['Country Name', '2022']]
population_2022.head()
```

```python
population_2022 = population_2022.dropna()
population_2022.head()
```

```python
top10 = population_2022.sort_values(by='2022', ascending=False).head(10)

top10
```

```python
exclude = [
    'World',
    'IDA & IBRD total',
    'IBRD only',
    'IDA total',
    'IDA blend',
    'IDA only',
    'Low & middle income',
    'Middle income',
    'Lower middle income',
    'Upper middle income',
    'High income',
    'Early-demographic dividend',
    'Late-demographic dividend',
    'Post-demographic dividend',
    'Pre-demographic dividend',
    'East Asia & Pacific',
    'Europe & Central Asia',
    'Latin America & Caribbean',
    'Middle East & North Africa',
    'North America',
    'South Asia',
    'Sub-Saharan Africa'
]

countries = population_2022[
    ~population_2022['Country Name'].isin(exclude)
]
```

```python
top10 = countries.sort_values(by='2022', ascending=False).head(10)

top10
```

## Data Visualization

A bar chart is created to compare the populations of the top 10 most populated countries in 2022.

```python
exclude_words = [
    "World","income","Europe","Asia","Africa","America",
    "Caribbean","OECD","IDA","IBRD","dividend",
    "countries","states","union","classification",
    "Euro area","European Union","Arab World"
]

countries = df.copy()

for word in exclude_words:
    countries = countries[
        ~countries["Country Name"].str.contains(word, case=False, na=False)
    ]

population_2022 = countries[['Country Name','2022']].dropna()

top10 = population_2022.sort_values(by='2022', ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.bar(top10['Country Name'], top10['2022'], color='steelblue')

plt.title('Top 10 Most Populated Countries (2022)', fontsize=16)
plt.xlabel('Country')
plt.ylabel('Population')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

## Observations

- India has the highest population in 2022.
- China is the second most populated country.
- There is a significant difference in population between the top two countries and the remaining countries.
- The bar chart provides a clear comparison of population distribution among the top ten countries.
